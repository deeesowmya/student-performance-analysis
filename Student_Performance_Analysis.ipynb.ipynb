```python
import pandas as pd
import numpy as np

np.random.seed(42)

students = []

departments = ["CSE", "IT", "ECE", "MECH", "CIVIL"]

for i in range(1, 51):
    students.append([
        1000 + i,
        f"Student_{i}",
        np.random.choice(departments),
        np.random.randint(60, 100),
        np.random.randint(40, 100),
        np.random.randint(1, 9)
    ])

df = pd.DataFrame(students, columns=[
    "Student ID",
    "Student Name",
    "Department",
    "Attendance Percentage",
    "Marks Obtained",
    "Semester"
])

df.to_csv("student_performance.csv", index=False)

print("Dataset Created Successfully!")
df.head()
```

```python
print("First 5 Records:")
print(df.head())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nSummary Statistics:")
print(df.describe())
```

```python
# Average Marks
print("Average Marks:", df["Marks Obtained"].mean())

# Highest Marks
print("Highest Marks:", df["Marks Obtained"].max())

# Lowest Marks
print("Lowest Marks:", df["Marks Obtained"].min())

# Department Wise Performance
print("\nDepartment Wise Average Marks:")
print(df.groupby("Department")["Marks Obtained"].mean())

# Attendance vs Marks Correlation
print("\nAttendance vs Marks Correlation:")
print(df["Attendance Percentage"].corr(df["Marks Obtained"]))

# Top 10 Students
print("\nTop 10 Students:")
print(df.sort_values(by="Marks Obtained", ascending=False).head(10))
```

```python
import matplotlib.pyplot as plt

dept_performance = df.groupby("Department")["Marks Obtained"].mean()

dept_performance.plot(kind="bar")

plt.title("Department Wise Average Marks")
plt.xlabel("Department")
plt.ylabel("Average Marks")
plt.show()
```

```python
df["Department"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Department Distribution")
plt.ylabel("")
plt.show()
```

```python
plt.hist(df["Marks Obtained"])

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```

```python
df_sorted = df.sort_values("Attendance Percentage")

plt.plot(
    df_sorted["Attendance Percentage"],
    df_sorted["Marks Obtained"]
)

plt.title("Attendance vs Marks")
plt.xlabel("Attendance Percentage")
plt.ylabel("Marks Obtained")
plt.show()
```
